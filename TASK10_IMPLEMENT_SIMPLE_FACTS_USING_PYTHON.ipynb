```python

from kanren import Relation, facts, var, run

# 1. Define a relation
teaches = Relation()

# 2. Add facts to the knowledge base
facts(
    teaches,
    ("Ravi", "Arun"),
    ("Priya", "Meena"),
    ("Ravi", "Kiran"),
    ("Suresh", "Divya")
)

# 3. Query: Who teaches Arun?
x = var()
result = run(1, x, teaches(x, "Arun"))

# 4. Display the result
print("The teacher of Arun is:", result[0])
```
