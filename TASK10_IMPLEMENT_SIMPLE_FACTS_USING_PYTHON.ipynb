```python

from kanren import Relation, facts, var, run

# 1. Define a relation
teaches = Relation()

# 2. Add facts to the knowledge base
facts(
    teaches,
    ("Ravi", "Arun"),
    ("Priya", "Meena"),
    ("Ravi", "Kiran"),
    ("Suresh", "Divya")
)

# 3. Query: Who teaches Arun?
x = var()
result = run(1, x, teaches(x, "Arun"))

# 4. Display the result
print("The teacher of Arun is:", result[0])
```

```python

from kanren import Relation, facts, var, run

# Define a relation
parent = Relation()

# Add facts: (Parent, Child)
facts(
    parent,
    ("Ramesh", "Kumar"),
    ("Lakshmi", "Kumar"),
    ("Suresh", "Anu"),
    ("Geetha", "Anu")
)

# Find the parent of Kumar
x = var()
result = run(0, x, parent(x, "Kumar"))

# Display the result
print("Parents of Kumar:")
for name in result:
    print(name)
```

```python

from kanren import Relation, facts, var, run

# Define a relation
likes = Relation()

# Add facts: (Person, Fruit)
facts(
    likes,
    ("Arun", "Apple"),
    ("Priya", "Mango"),
    ("Kiran", "Banana"),
    ("Divya", "Orange")
)

# Find the fruit liked by Priya
x = var()
result = run(1, x, likes("Priya", x))

# Display the result
print("Priya likes:", result[0])
```
