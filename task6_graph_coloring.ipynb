```python
def is_safe(v, graph, color, c):
    for u in range(len(graph)):
        if graph[v][u] == 1 and color[u] == c:
            return False
    return True

def graph_coloring(graph, m, color, v):
    n = len(graph)

    if v == n:
        return True

    for c in range(1, m + 1):
        if is_safe(v, graph, color, c):
            color[v] = c

            if graph_coloring(graph, m, color, v + 1):
                return True

            color[v] = 0

    return False

graph = [
    [0, 1, 0, 1, 0],  # A
    [1, 0, 0, 1, 0],  # B
    [0, 0, 0, 1, 0],  # C
    [1, 1, 1, 0, 1],  # D
    [0, 0, 0, 1, 0]   # E
]

vertices = ['A', 'B', 'C', 'D', 'E']

for m in range(1, len(graph) + 1):
    color = [0] * len(graph)

    if graph_coloring(graph, m, color, 0):
        print("Minimum number of colors:", m)
        for i in range(len(vertices)):
            print(vertices[i], "-> Color", color[i])
        break
```
