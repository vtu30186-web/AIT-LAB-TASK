```python
def monkey_banana():
    state = {"monkey": "floor", "box": "window", "on_box": False}
    print("Monkey is on floor")
    state["monkey"] = "window"
    print("Monkey moves to box")
    state["box"] = "center"
    print("Monkey pushes box to center")
    state["on_box"] = True
    print("Monkey climbs box")
    if state["on_box"] and state["box"] == "center":
        print("Monkey touches bananas!")
        print("Goal Reached!")

monkey_banana()
```
