```python
import heapq
def a_star_algorithm(graph, heuristics, start, goal):
    open_set = []
    heapq.heappush(open_set, (heuristics[start], start))
    g_score = {node: float('inf') for node in graph}
    g_score[start] = 0
    came_from = {}
    while open_set:
        current_f, current = heapq.heappop(open_set)
        if current == goal:
            path = []
            while current in came_from:
                path.append(current)
                current = came_from[current]
            path.append(start)
            return path[::-1]
        for neighbour, weight in graph[current]:
            tentative_g = g_score[current] + weight
            if tentative_g < g_score[neighbour]:
                came_from[neighbour] = current
                g_score[neighbour] = tentative_g
                f_score = tentative_g + heuristics[neighbour]
                heapq.heappush(open_set, (f_score, neighbour))
    return None
my_graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('A', 1), ('C', 2), ('D', 5)],
    'C': [('A', 4), ('B', 2), ('D', 1)],
    'D': [('B', 5), ('C', 1)]
}
my_heuristics = {
    'A': 4,
    'B': 2,
    'C': 1,
    'D': 0
}
shortest_path = a_star_algorithm(
    my_graph,
    my_heuristics,
    start='A',
    goal='D'
)
print("The optimal path found by A* is:", shortest_path)
```
